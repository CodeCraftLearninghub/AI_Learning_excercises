### 📘 Notes — Splitting Data into Training and Testing Sets

`train_test_split()` divides the dataset into separate portions.

- **Training data** → used by the model to learn patterns.
- **Testing data** → kept aside to evaluate how the trained model performs on unseen data.
- `test_size` controls how much data is reserved for testing.
- `random_state` makes the split reproducible, so students can obtain the same split when running the notebook again.

**Student idea:** We should not judge a model only on the data it already saw during training.

```python
X_train, X_test, y_train, y_test = train_test_split(df[['age']],df.bought_insurance,train_size=0.8)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

### 📘 Notes — Training the Model

`model.fit(X, y)` trains the machine-learning model.

- `X` contains the input features.
- `y` contains the known target/class labels.
- During `fit()`, Logistic Regression learns parameters that help separate the classes.

**Student idea:** `fit()` means **learn from the training data**.

After this step, the model is ready to make predictions for new data.

```python
model.fit(X_train, y_train)
```

### 📘 Notes — Making a Classification Prediction

`model.predict(...)` asks the trained Logistic Regression model to predict the class for new input data.

For example, the result may be:

- `0` → predicted negative class
- `1` → predicted positive class

The exact meaning of `0` and `1` depends on the dataset.

**Important:** The new input must have the same feature structure used during training.

If the model was trained with a DataFrame containing named columns, it is best to predict using a DataFrame with the same column names. This avoids the Scikit-learn warning about feature names.

Example:

`model.predict(pd.DataFrame({'Hours': [80]}))`

instead of:

`model.predict([[80]])`

```python
y_predicted = model.predict(X_test)
```
